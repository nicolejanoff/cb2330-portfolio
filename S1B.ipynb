```python
x = [2, 5, 3, 4] #Listan jag vill räkna på

running_total = [] #Skapar en tom lista där jag vill spara resultaten jag räknat ut

total = 0 #Skapar en accumulator. Den är som en räknare/samlare, =0 eftersom vi ännu inte har lagt till något

for value in x: #Ta ett värde taget från x och kalla det value
  total = total + value #Samma som att man skulle skriva total += value
  running_total.append(total) #Lägg till i slutet av listan

print(running_total)
```

```python
sequence = "ATGAAGCCAGA"

gc_count = 0 #Accumulator

for base in sequence: #Går igenom en bas i taget
  if base == "G" or base == "C": #Är denna bas G eller C?
    gc_count += 1 #Om svar ja så ökar vi räknaren med 1

gc_content = gc_count / len(sequence)
print(gc_content)
```

```python
def factorial(n):
  result = 1

  for i in range(n, n+1):
    result = result * i

  return result

def binomial_coefficient(n, k):
  return factorial(n) / factorial(k)*factorial(n-k)

print(binomial_coefficient(11,2))
```
